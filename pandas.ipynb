```
import pandas as pd
import numpy as np
```

```
df = pd.read_csv(
    "QS_World_University_Rankings_2025_Top_global_universities (1).csv",
    encoding="latin1"
)
```

```
df.isnull().sum().sum()
```

```
df.isnull().sum()
```

```
df.dtypes
```

```
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:

    df[col] = df[col].fillna(
        df[col].mean()
    )
```

```
text_cols = df.select_dtypes(include="object").columns

for col in text_cols:

    df[col] = df[col].fillna(
        df[col].mode()[0]
    )
```

```
df = df.drop_duplicates()
```

```
for col in text_cols:

    df[col] = df[col].str.strip()
```

```
df.isnull().sum().sum()
```

```
df.info()
```

```
df.describe()
```

```
df.isnull().sum().sum()
```

```
df.dtypes
```

```
for col in df.columns:

    try:
        df[col] = pd.to_numeric(df[col])

    except:
        pass

print(df.dtypes)
```

```
numeric_cols = df.select_dtypes(include=["float64"]).columns

for col in numeric_cols:

    df[col] = df[col].round().astype(int)

print(df.dtypes)
```

```
df.dtypes
```

```
for col in df.columns:

    try:

        # remove commas if present
        df[col] = df[col].astype(str).str.replace(",", "")

        # convert to numeric
        df[col] = pd.to_numeric(df[col])

        # convert float to int
        df[col] = df[col].fillna(0).astype(int)

    except:
        pass
print(df.dtypes)
```

```
# ==========================================
# CLEAN RANK COLUMNS
# ==========================================

rank_cols = [
    "RANK_2025",
    "RANK_2024",
    "Academic_Reputation_Rank",
    "Employer_Reputation_Rank",
    "Faculty_Student_Rank",
    "Citations_per_Faculty_Rank",
    "International_Faculty_Rank",
    "International_Students_Rank",
    "International_Research_Network_Rank",
    "Employment_Outcomes_Rank",
    "Sustainability_Rank",
    "Institution_Name",
    "Location",
    "Region",
    "SIZE",
    "FOCUS",
    "RES.",
    "STATUS"]



for col in rank_cols:

    # keep only numbers
    df[col] = df[col].astype(str).str.extract(r'(\d+)')

    # convert to float
    df[col] = pd.to_numeric(df[col], errors="coerce")

    # fill missing
    df[col] = df[col].fillna(0)

    # convert to integer
    df[col] = df[col].astype(int)



# ==========================================
# CHECK DATA TYPES
# ==========================================

print(df.dtypes)
```
