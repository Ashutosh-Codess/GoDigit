```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
plt.plot(x, y)
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
plt.plot(x, y)
plt.title("Student performance")
plt.xlabel("Exam Number")
plt.ylabel("Marks")

plt.show()
```

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]

plt.plot(
    x,
    y,
    color = 'red',
    marker = 'o',
    linestyle = '-',
    linewidth = 3
)
plt.title("Coustomized performance")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
x = [1, 2, 3, 4]

maths = [50, 60, 70, 80]
science = [45, 55, 75, 85]

plt.plot(x, maths, label = "Maths")
plt.plot(x, science, label = 'Science')

plt.title("Marks Conparison")
plt.xlabel("tests")
plt.ylabel("marks")

plt.legend()
plt.show()
```
