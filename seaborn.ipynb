```python
import pandas as pd

df = pd.read_csv(
    "dataset.csv",
    encoding='latin1'
)
```

```python
df['Overall_Score'] = pd.to_numeric(
    df['Overall_Score'],
    errors='coerce'
)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(
    x='Overall_Score',
    y='Academic_Reputation_Score',
    data=df
)

plt.title("Overall Score vs Academic Reputation")
plt.xlabel("Overall Score")
plt.ylabel("Academic Reputation Score")

plt.show()
```

```python
sns.histplot(
    df['Overall_Score'].dropna()
)

plt.title("Distribution of Overall Score")

plt.show()
```

```python
sns.boxplot(
    x=df['Overall_Score']
)

plt.title("Outlier Detection")

plt.show()
```

```python
top10 = df.head(10)

plt.figure(figsize=(12,5))

sns.barplot(
    x='Institution_Name',
    y='Overall_Score',
    data=top10
)

plt.xticks(rotation=90)

plt.title("Top 10 Universities")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='Region',
    data=df
)

plt.title("University Count by Region")

plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(12,8))

sns.heatmap(
    numeric_df.corr(),
    annot=True
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
sns.pairplot(
    df[
        [
            'Overall_Score',
            'Academic_Reputation_Score',
            'Employer_Reputation_Score'
        ]
    ]
)

plt.show()
```
