```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]

sales = [10,20,15,30,25]
profit = [5,10,8,15,12]

plt.plot(x, sales, label='Sales')
plt.plot(x, profit, label='Profit')

plt.title("Sales vs Profit")
plt.xlabel("Months")
plt.ylabel("Amount")

plt.legend()

plt.show()
```
