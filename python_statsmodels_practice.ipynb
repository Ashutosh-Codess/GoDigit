```


import pandas as pd
import numpy as np




df = pd.read_csv(
    "dataset.csv",
    encoding="latin1"
)



print("\nFIRST 5 ROWS")
print(df.head())

print("\nLAST 5 ROWS")
print(df.tail())

print("\nDATASET SHAPE")
print(df.shape)

print("\nCOLUMN NAMES")
print(df.columns)

print("\nDATA TYPES")
print(df.dtypes)

print("\nFULL INFORMATION")
print(df.info())



print("\nMISSING VALUES IN EACH COLUMN")
print(df.isnull().sum())

print("\nTOTAL MISSING VALUES")
print(df.isnull().sum().sum())




print("\nTOTAL DUPLICATE ROWS")
print(df.duplicated().sum())




df = df.drop_duplicates()

print("\nDUPLICATES REMOVED")





df.columns = df.columns.str.strip()


text_cols = df.select_dtypes(include=["object", "string"]).columns

for col in text_cols:
    df[col] = df[col].astype(str).str.strip()

print("\nEXTRA SPACES REMOVED")




numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())



# text columns

text_cols = df.select_dtypes(include=["object", "string"]).columns

for col in text_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nMISSING VALUES FILLED")





print("\nMISSING VALUES AFTER CLEANING")
print(df.isnull().sum())

print("\nTOTAL MISSING VALUES AFTER CLEANING")
print(df.isnull().sum().sum())




for col in df.columns:

    try:
        df[col] = pd.to_numeric(df[col])

    except:
        pass

print("\nPOSSIBLE NUMERIC COLUMNS CONVERTED")




print("\nFINAL DATASET INFORMATION")
print(df.info())

print("\nFINAL DATA TYPES")
print(df.dtypes)

print("\nFINAL SHAPE")
print(df.shape)




print("\nSTATISTICAL SUMMARY")
print(df.describe())





df.to_csv(
    "cleaned_dataset.csv",
    index=False
)

print("\nCLEANED DATASET SAVED SUCCESSFULLY")
```

```
import pandas as pd
from scipy.stats import ttest_1samp

# Load dataset with encoding fix
df = pd.read_csv(
    "dataset.csv",
    encoding="latin1"
)

# Convert to numeric
df["Overall_Score"] = pd.to_numeric(
    df["Overall_Score"],
    errors="coerce"
)

# Remove missing values
score = df["Overall_Score"].dropna()

# One Sample T-Test
t_stat, p_value = ttest_1samp(
    score,
    50
)

# Print results
print("T-statistic:", t_stat)
print("P-value:", p_value)

# Interpretation
if p_value < 0.05:
    print("Significant difference exists from 50")
else:
    print("No significant difference from 50")
```

```
usa = df[df["Location"] == "United States"]["Academic_Reputation_Score"]

uk = df[df["Location"] == "United Kingdom"]["Academic_Reputation_Score"]
```

```
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(
    usa,
    uk,
    nan_policy="omit"
)

print("T-statistic:", t_stat)

print("P-value:", p_value)
```
